### Decisiones a partir de EDA:

Numericas:

Todas las features contienen una proporción grande de outliers, por lo tanto se aplicara transformación logarítmica y después estandarización

Categóricas:

En funcion de la cardinalidad y naturaleza, se decidió aplicar

    - workclass (Cardinalidad menor a 10)
    - marital-status (Cardinalidad menor a 10)

    - relationship (Cardinalidad menor a 10)
    - race (Cardinalidad menor a 10)
- Binary:
    - sex (solo 2 posibles valores)
- Ordinal:
    - education (Se puede ordenar numericamente sin perder sentido)
- Embeddings:
    - skill-profile (Cardinalidad mucho mayor a 10)
    - native-country (Cardinalidad mucho mayor a 10)


```python
num_cols = [
    'age',
    'capital-gain',
    'capital-loss',
    'hours-per-week',
]
ohe_cols = [
    'workclass',
    'marital-status',
    'occupation',
    'relationship',
    'race',
]
binary_cols = ['sex']
ordinal_cols = ['education']
embedding_cols = ['skill-profile', 'native-country']

education_order = [
    'Preschool',
    '1st-4th',
    '5th-6th',
    '7th-8th',
    '9th',
    '10th',
    '11th',
    '12th',
    'HS-grad',
    'Some-college',
    'Assoc-voc',
    'Assoc-acdm',
    'Bachelors',
    'Masters',
    'Prof-school',
    'Doctorate',
]
```

```python
num_pipeline = Pipeline(
    [
        ('log1p',FunctionTransformer(np.log1p, validate=False,feature_names_out='one-to-one')),  # log(1+x) tolera ceros
        ('scaler', StandardScaler()),
    ]
)
ordinal_pipeline = OrdinalEncoder(
    categories=[education_order],
    handle_unknown='use_encoded_value',unknown_value=-1,
)

preprocessor = ColumnTransformer(
    transformers=[
        ('num', num_pipeline, num_cols),
        ('ohe', OneHotEncoder(handle_unknown='ignore',sparse_output=False), ohe_cols),
        ('binary', OneHotEncoder(drop='first',sparse_output=False), binary_cols),
        ('ordinal', ordinal_pipeline, ordinal_cols),
        ('emb', OrdinalEncoder(), embedding_cols),
    ],
    remainder='drop',
).set_output(transform='pandas')
```

```python
X_train = df.drop(columns=['income'])
y_train = df['income']
X_val = df_val.drop(columns='income')
y_val = df_val['income']

X_train = preprocessor.fit_transform(X_train)
X_val = preprocessor.transform(X_val)
```

## Modelo con embeddings

```python
class IncomeDS(Dataset):
    def __init__(self, X, y):
        self.skill = torch.tensor(X['emb__skill-profile'].values, dtype=torch.long)
        self.county = torch.tensor(X['emb__native-country'].values, dtype=torch.long)
        self.rest = torch.tensor(
            X.drop(columns=['emb__native-country', 'emb__skill-profile']).values,
            dtype=torch.float32,
        )
        y_numeric = (y.astype(str).str.strip().str.contains('>50K')).values.astype(np.float32)
        self.income = torch.tensor(y_numeric, dtype=torch.float32)
    def __len__(self):
        return len(self.income)
    def __getitem__(self, i):
        return self.rest[i], self.skill[i], self.county[i], self.income[i]
```

```python
class IncomeModel(nn.Module):

  def __init__(
      self, n_rest, n_skill, n_country, skill_dim, country_dim, drop_p=0.3
  ):
    super().__init__()
    self.skill_emb = nn.Embedding(n_skill, skill_dim)
    self.country_emb = nn.Embedding(n_country, country_dim)
    nn.init.normal_(self.skill_emb.weight, std=0.01)
    nn.init.normal_(self.country_emb.weight, std=0.01)
    self.dropout = nn.Dropout(drop_p)

    # Input dim = n_rest + skill_dim + country_dim
    self.total_input_dim = n_rest + skill_dim + country_dim

    self.mlp = nn.Sequential(
        nn.Linear(self.total_input_dim, 256),
        nn.LeakyReLU(),
        nn.Dropout(drop_p),
        nn.Linear(256, 128),
        nn.LeakyReLU(),
        nn.Dropout(drop_p),
        nn.Linear(128, 64),
        nn.LeakyReLU(),
        nn.Dropout(drop_p),
        nn.Linear(64, 1),
        nn.Sigmoid(),
    )

  def forward(self, rest, skill, country):
    se = self.skill_emb(skill)
    ce = self.country_emb(country)
    x = torch.cat([rest, se, ce], dim=1)
    x = self.dropout(x)
    out = self.mlp(x)
    return out.squeeze(1)
```

```python
train_ds = IncomeDS(X_train, y_train)
val_ds = IncomeDS(X_val, y_val)

batch_size = 256
train_loader = DataLoader(
    train_ds, batch_size=batch_size, shuffle=True, drop_last=False
)
val_loader = DataLoader(
    val_ds, batch_size=batch_size, shuffle=False, drop_last=False
)

# Cardinalidad
n_skill = int(X_train['emb__skill-profile'].max()) + 1
n_country = int(X_train['emb__native-country'].max()) + 1
n_rest = X_train.shape[1] - 2

# Criterio para dimension de embeddings fastai rule 
skill_dim = int(min(50, n_skill**0.25 + 1))
country_dim = int(min(50, n_country**0.25 + 1))
print(
f"""
Dimension de skill -> {skill_dim}
Dimension de country -> {country_dim} 
"""
)

device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
model = IncomeModel(
    n_rest, n_skill, n_country, skill_dim, country_dim, drop_p=0.3
).to(device)

criterion = nn.BCELoss().to(device)
f1_macro = F1Score(task='multiclass', num_classes=2, average='macro').to(device)
acc = BinaryAccuracy().to(device)

optimizer = torch.optim.Adam(model.parameters(), lr=1e-3)

best_val = float('inf')
best_model_weights = None
epochs = 30
train_bce_hist = []
val_bce_hist = []
train_f1_hist = []
val_f1_hist = []
train_acc_hist = []
val_acc_hist = []

for epoch in range(0, epochs):
    model.train()
    train_losses = []
    train_f1s = []
    train_accs = []
    
    for r, s, c, y in train_loader:
        r, s, c, y = r.to(device), s.to(device), c.to(device), y.to(device)
        pred = model(r, s, c)
        loss = criterion(pred, y)

        optimizer.zero_grad()
        loss.backward()
        optimizer.step()
        train_losses.append(loss.item())

        pred_labels = (pred > 0.5).long()

        f1_train = f1_macro(pred_labels, y.long())
        acc_train = acc(pred, y)
        train_f1s.append(f1_train.item())
        train_accs.append(acc_train.item())

    train_bce = float(np.mean(train_losses))
    train_f1 = float(np.mean(train_f1s))
    train_acc = float(np.mean(train_accs))

    model.eval()
    ys, ps = [], []
    with torch.no_grad():
        for r, s, c, y in val_loader:
            r, s, c, y = r.to(device), s.to(device), c.to(device), y.to(device)
            p = model(r, s, c)
            ys.append(y)
            ps.append(p)

        y_true = torch.cat(ys)
        p_pred = torch.cat(ps)
        val_bce = criterion(p_pred, y_true).item()
        
        p_pred_labels = (p_pred > 0.5).long()
        val_f1 = f1_macro(p_pred_labels, y_true.long()).item()
        val_acc = acc(p_pred, y_true).item()

    train_bce_hist.append(train_bce)
    train_f1_hist.append(train_f1)
    train_acc_hist.append(train_acc)
    val_bce_hist.append(val_bce)
    val_f1_hist.append(val_f1)
    val_acc_hist.append(val_acc)
    
    print(
        f'Epoch {epoch:02d} | train_bce={train_bce:.4f} | val_bce={val_bce:.4f} | '
        f'train_f1={train_f1:.4f} | val_f1={val_f1:.4f}'
    )

    if val_bce < best_val:
        best_val = val_bce
        best_model_weights = copy.deepcopy(model.state_dict())

```

```python
# BCE, F1, Acc Graficas por epoca

epochs = range(0, len(train_bce_hist))
best_bce_epoch = np.argmin(val_bce_hist)  # Minimum loss
best_f1_epoch = np.argmax(val_f1_hist)   # Maximum F1 score
best_acc_epoch = np.argmax(val_acc_hist)  # Maximum Accuracy

# colors
train_color = '#1f77b4'  
val_color = '#ff7f0e'    
v_line_color = '#555555'

fig, axes = plt.subplots(1, 3, figsize=(18, 5))

# ---------------- 1. BCE Loss Subplot ----------------
axes[0].plot(epochs, train_bce_hist, label='Train BCE', color=train_color, linewidth=2)
axes[0].plot(epochs, val_bce_hist, label='Val BCE', color=val_color, linewidth=2)
axes[0].axvline(
    x=best_bce_epoch, color=v_line_color, linestyle='--', linewidth=1.5,
    label=f'Best Val (Ep {best_bce_epoch}: {val_bce_hist[best_bce_epoch-1]:.4f})'
)
axes[0].set_title('BCE Loss', fontsize=14, fontweight='bold')
axes[0].set_xlabel('Epoch', fontsize=12)
axes[0].set_ylabel('Loss', fontsize=12)
axes[0].legend(fontsize=10, loc='best')
axes[0].grid(True, linestyle=':', alpha=0.6)

# ---------------- 2. Macro F1-Score Subplot ----------------
axes[1].plot(epochs, train_f1_hist, label='Train F1', color=train_color, linewidth=2)
axes[1].plot(epochs, val_f1_hist, label='Val F1', color=val_color, linewidth=2)
axes[1].axvline(
    x=best_f1_epoch, color=v_line_color, linestyle='--', linewidth=1.5,
    label=f'Best Val (Ep {best_f1_epoch}: {val_f1_hist[best_f1_epoch-1]:.4f})'
)
axes[1].set_title('Macro F1-Score', fontsize=14, fontweight='bold')
axes[1].set_xlabel('Epoch', fontsize=12)
axes[1].set_ylabel('F1-Score', fontsize=12)
axes[1].legend(fontsize=10, loc='best')
axes[1].grid(True, linestyle=':', alpha=0.6)

# ---------------- 3. Accuracy Subplot ----------------
axes[2].plot(epochs, train_acc_hist, label='Train Acc', color=train_color, linewidth=2)
axes[2].plot(epochs, val_acc_hist, label='Val Acc', color=val_color, linewidth=2)
axes[2].axvline(
    x=best_acc_epoch, color=v_line_color, linestyle='--', linewidth=1.5,
    label=f'Best Val (Ep {best_acc_epoch}: {val_acc_hist[best_acc_epoch-1]:.4f})'
)
axes[2].set_title('Accuracy', fontsize=14, fontweight='bold')
axes[2].set_xlabel('Epoch', fontsize=12)
axes[2].set_ylabel('Accuracy', fontsize=12)
axes[2].legend(fontsize=10, loc='best')
axes[2].grid(True, linestyle=':', alpha=0.6)

plt.tight_layout()
plt.show()

```

```python
# Matriz de confusion
if best_model_weights is not None:
    model.load_state_dict(best_model_weights)
model.eval()
ys, ps = [], []

with torch.no_grad():
  for r, s, c, y in val_loader:
    r, s, c = r.to(device), s.to(device), c.to(device)
    p = model(r, s, c)
    ys.append(y.cpu())
    ps.append(p.cpu())

y_true = torch.cat(ys).numpy()
p_pred = torch.cat(ps).numpy()

# Convertir a 0 o 1
y_pred = (p_pred >= 0.5).astype(int)
y_true = y_true.astype(int)

# Matrices
cm = confusion_matrix(y_true, y_pred)
cm_norm = confusion_matrix(y_true, y_pred, normalize='true')
labels = ['<=50K', '>50K']

fig, axes = plt.subplots(1, 2, figsize=(13, 5))

sns.heatmap(
    cm,
    annot=True,
    fmt='d',
    cmap='Blues',
    xticklabels=labels,
    yticklabels=labels,
    cbar=False,
    annot_kws={'size': 14, 'weight': 'bold'},
    ax=axes[0]
)
axes[0].set_title('Matriz de Confusion (Absoluta)', fontsize=14, pad=12)
axes[0].set_xlabel('Predicted', fontsize=12)
axes[0].set_ylabel('True', fontsize=12)

sns.heatmap(
    cm_norm,
    annot=True,
    fmt='.2',
    cmap='Blues',
    xticklabels=labels,
    yticklabels=labels,
    cbar=False,
    annot_kws={'size': 14, 'weight': 'bold'},
    ax=axes[1]
)
axes[1].set_title('Matriz de Confusion (Relativa)', fontsize=14, pad=12)
axes[1].set_xlabel('Predicted', fontsize=12)
axes[1].set_ylabel('True', fontsize=12)

plt.tight_layout()
plt.show()
```

## Modelo sin embeddings

```python
num_cols = [
    'age',
    'capital-gain',
    'capital-loss',
    'hours-per-week',
]
ohe_cols = [
    'workclass',
    'marital-status',
    'occupation',
    'relationship',
    'race',
    'skill-profile',
    'native-country'
]
binary_cols = ['sex']
ordinal_cols = ['education']
education_order = [
    'Preschool',
    '1st-4th',
    '5th-6th',
    '7th-8th',
    '9th',
    '10th',
    '11th',
    '12th',
    'HS-grad',
    'Some-college',
    'Assoc-voc',
    'Assoc-acdm',
    'Bachelors',
    'Masters',
    'Prof-school',
    'Doctorate',
]
```

```python
num_pipeline = Pipeline(
    [
        ('log1p',FunctionTransformer(np.log1p, validate=False,feature_names_out='one-to-one')),  # log(1+x) tolera ceros
        ('scaler', StandardScaler()),
    ]
)
ordinal_pipeline = OrdinalEncoder(
    categories=[education_order],
    handle_unknown='use_encoded_value',unknown_value=-1,
)

preprocessor = ColumnTransformer(
    transformers=[
        ('num', num_pipeline, num_cols),
        ('ohe', OneHotEncoder(handle_unknown='ignore',sparse_output=False), ohe_cols),
        ('binary', OneHotEncoder(drop='first',sparse_output=False), binary_cols),
        ('ordinal', ordinal_pipeline, ordinal_cols),
    ],
    remainder='drop',
).set_output(transform='pandas')
```

```python
X_train = df.drop(columns=['income'])
y_train = df['income']
X_val = df_val.drop(columns='income')
y_val = df_val['income']

X_train = preprocessor.fit_transform(X_train)
X_val = preprocessor.transform(X_val)
```

```python
class IncomeDS(Dataset):
    def __init__(self, X, y):
        self.rest = torch.tensor(
            X.values,
            dtype=torch.float32,
        )
        y_numeric = (y.astype(str).str.strip().str.contains('>50K')).values.astype(np.float32)
        self.income = torch.tensor(y_numeric, dtype=torch.float32)
    def __len__(self):
        return len(self.income)
    def __getitem__(self, i):
        return self.rest[i], self.income[i]
```

```python
class IncomeModel(nn.Module):

  def __init__(
      self, n_rest
  ):
    super().__init__()

    self.total_input_dim = n_rest

    self.mlp = nn.Sequential(
        nn.Linear(self.total_input_dim, 256),
        nn.LeakyReLU(),
        nn.Linear(256, 128),
        nn.LeakyReLU(),
        nn.Linear(128, 64),
        nn.LeakyReLU(),
        nn.Linear(64, 1),
        nn.Sigmoid(),
    )

  def forward(self, rest):
    x = torch.cat([rest], dim=1)
    out = self.mlp(x)
    return out.squeeze(1)
```

```python
train_ds = IncomeDS(X_train, y_train)
val_ds = IncomeDS(X_val, y_val)

batch_size = 256
train_loader = DataLoader(
    train_ds, batch_size=batch_size, shuffle=True, drop_last=False
)
val_loader = DataLoader(
    val_ds, batch_size=batch_size, shuffle=False, drop_last=False
)

# Cardinalidad
n_rest = X_train.shape[1]


device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
model = IncomeModel(
    n_rest
).to(device)

criterion = nn.BCELoss().to(device)

f1_macro = F1Score(task='multiclass', num_classes=2, average='macro').to(device)
acc = BinaryAccuracy().to(device)
optimizer = torch.optim.Adam(model.parameters(), lr=1e-3)

best_val = float('inf')
best_model_weights = None
epochs = 30
train_bce_hist = []
val_bce_hist = []
train_f1_hist = []
val_f1_hist = []
train_acc_hist = []
val_acc_hist = []

for epoch in range(epochs):
    model.train()
    train_losses = []
    train_f1s = []
    train_accs = []
    
    for r, y in train_loader:
        r, y = r.to(device), y.to(device)
        pred = model(r)
        loss = criterion(pred, y)

        optimizer.zero_grad()
        loss.backward()
        optimizer.step()
        train_losses.append(loss.item())

        pred_labels = (pred > 0.5).long()

        f1_train = f1_macro(pred_labels, y.long())
        acc_train = acc(pred, y)
        train_f1s.append(f1_train.item())
        train_accs.append(acc_train.item())

    train_bce = float(np.mean(train_losses))
    train_f1 = float(np.mean(train_f1s))
    train_acc = float(np.mean(train_accs))

    model.eval()
    ys, ps = [], []
    with torch.no_grad():
        for r, y in val_loader:
            r, y = r.to(device), y.to(device)
            p = model(r)
            ys.append(y)
            ps.append(p)

        y_true = torch.cat(ys)
        p_pred = torch.cat(ps)
        val_bce = criterion(p_pred, y_true).item()
        
        p_pred_labels = (p_pred > 0.5).long()
        val_f1 = f1_macro(p_pred_labels, y_true.long()).item()
        val_acc = acc(p_pred, y_true).item()

    train_bce_hist.append(train_bce)
    train_f1_hist.append(train_f1)
    train_acc_hist.append(train_acc)
    val_bce_hist.append(val_bce)
    val_f1_hist.append(val_f1)
    val_acc_hist.append(val_acc)
    
    print(
        f'Epoch {epoch:02d} | train_bce={train_bce:.4f} | val_bce={val_bce:.4f} | '
        f'train_f1={train_f1:.4f} | val_f1={val_f1:.4f}'
    )

    if val_bce < best_val:
        best_val = val_bce
        best_model_weights = copy.deepcopy(model.state_dict())
```

```python
# Graficos de BCE, F1, Acc
epochs = range(0, len(train_bce_hist))

best_bce_epoch = np.argmin(val_bce_hist)  
best_f1_epoch = np.argmax(val_f1_hist)
best_acc_epoch = np.argmax(val_acc_hist)  

# colors
train_color = '#1f77b4'  
val_color = '#ff7f0e'    
v_line_color = '#555555' 

fig, axes = plt.subplots(1, 3, figsize=(18, 5))

# ---------------- 1. BCE Loss Subplot ----------------
axes[0].plot(epochs, train_bce_hist, label='Train BCE', color=train_color, linewidth=2)
axes[0].plot(epochs, val_bce_hist, label='Val BCE', color=val_color, linewidth=2)
axes[0].axvline(
    x=best_bce_epoch, color=v_line_color, linestyle='--', linewidth=1.5,
    label=f'Best Val (Ep {best_bce_epoch}: {val_bce_hist[best_bce_epoch-1]:.4f})'
)
axes[0].set_title('BCE Loss', fontsize=14, fontweight='bold')
axes[0].set_xlabel('Epoch', fontsize=12)
axes[0].set_ylabel('Loss', fontsize=12)
axes[0].legend(fontsize=10, loc='best')
axes[0].grid(True, linestyle=':', alpha=0.6)

# ---------------- 2. Macro F1-Score Subplot ----------------
axes[1].plot(epochs, train_f1_hist, label='Train F1', color=train_color, linewidth=2)
axes[1].plot(epochs, val_f1_hist, label='Val F1', color=val_color, linewidth=2)
axes[1].axvline(
    x=best_f1_epoch, color=v_line_color, linestyle='--', linewidth=1.5,
    label=f'Best Val (Ep {best_f1_epoch}: {val_f1_hist[best_f1_epoch-1]:.4f})'
)
axes[1].set_title('Macro F1-Score', fontsize=14, fontweight='bold')
axes[1].set_xlabel('Epoch', fontsize=12)
axes[1].set_ylabel('F1-Score', fontsize=12)
axes[1].legend(fontsize=10, loc='best')
axes[1].grid(True, linestyle=':', alpha=0.6)

# ---------------- 3. Accuracy Subplot ----------------
axes[2].plot(epochs, train_acc_hist, label='Train Acc', color=train_color, linewidth=2)
axes[2].plot(epochs, val_acc_hist, label='Val Acc', color=val_color, linewidth=2)
axes[2].axvline(
    x=best_acc_epoch, color=v_line_color, linestyle='--', linewidth=1.5,
    label=f'Best Val (Ep {best_acc_epoch}: {val_acc_hist[best_acc_epoch-1]:.4f})'
)
axes[2].set_title('Accuracy', fontsize=14, fontweight='bold')
axes[2].set_xlabel('Epoch', fontsize=12)
axes[2].set_ylabel('Accuracy', fontsize=12)
axes[2].legend(fontsize=10, loc='best')
axes[2].grid(True, linestyle=':', alpha=0.6)

plt.tight_layout()
plt.show()
```

```python
if best_model_weights is not None:
    model.load_state_dict(best_model_weights)
model.eval()
ys, ps = [], []

with torch.no_grad():
  for r, y in val_loader:
    r = r.to(device)
    p = model(r)
    ys.append(y.cpu())
    ps.append(p.cpu())

y_true = torch.cat(ys).numpy()
p_pred = torch.cat(ps).numpy()

# Convertir a 0 o 1
y_pred = (p_pred >= 0.5).astype(int)
y_true = y_true.astype(int)

# Matrices de confusion
cm = confusion_matrix(y_true, y_pred)
cm_norm = confusion_matrix(y_true, y_pred, normalize='true')
labels = ['<=50K', '>50K']

fig, axes = plt.subplots(1, 2, figsize=(13, 5))

sns.heatmap(
    cm,
    annot=True,
    fmt='d',
    cmap='Blues',
    xticklabels=labels,
    yticklabels=labels,
    cbar=False,
    annot_kws={'size': 14, 'weight': 'bold'},
    ax=axes[0]
)
axes[0].set_title('Matriz de Confusion (Absoluta)', fontsize=14, pad=12)
axes[0].set_xlabel('Predicted', fontsize=12)
axes[0].set_ylabel('True', fontsize=12)

sns.heatmap(
    cm_norm,
    annot=True,
    fmt='.2',
    cmap='Blues',
    xticklabels=labels,
    yticklabels=labels,
    cbar=False,
    annot_kws={'size': 14, 'weight': 'bold'},
    ax=axes[1]
)
axes[1].set_title('Matriz de Confusion (Relativa)', fontsize=14, pad=12)
axes[1].set_xlabel('Predicted', fontsize=12)
axes[1].set_ylabel('True', fontsize=12)

plt.tight_layout()
plt.show()
```

## Conclusiones

Para determinar cual de los modelos es mejor, es necesario considerar distintos ejes. Para comenzar, comparamos las metricas obtenidas en ambos.

| Model | Precision (Macro) | Recall (Macro) | F1-Score (Macro) |
| :--- | :---: | :---: | :---: |
| **Sin Embeddings** | 0.8538 | 0.8428 | 0.8481 |
| **Con Embeddings** | 0.8532 | 0.8361 | 0.8441 |

Observamos que los resultados son muy similares, por lo que no estamos en condiciones de asegurar que uno es mejor que otro en este aspecto.

Otro eje de comparación reside en las dimensiones de los mismos

| Modelo | Dimensiones de Entrada |
| :--- | :---: |
| **Sin Embeddings** | 166 |
| **Con Embeddings** | 51 |

Aquí el modelo con embeddings reduce en aproximadamente 3 la dimension de las entradas. Esto se traducirá en menos parametros para entrenar y menos operaciones para hacer inferencia en la primer capa. Nuestro modelo al contar con 128 neuronas en la primer capa oculta, comparamos la cantidad de pesos que tendremos en cada modelo para hacer explicita la diferencia.

| Configuración de Entrada | Dimensiones de Entrada | Neuronas Capa Oculta | Cálculo de Pesos | Pesos Totales (Capa 1) |
| :--- | :---: | :---: | :---: | :---: |
| **Sin Embeddings (One-Hot)** | 166 | 128 | $166 \times 128$ | 21,248 |
| **Con Embeddings** | 51 | 128 | $51 \times 128$ | 6,528 |

Esta diferencia nos permite decir que el modelo con embeddings es mejor para este problema en particular, ya que sin sacrificar rendimiento logro reducir la cantidad de parametros necesarios.
